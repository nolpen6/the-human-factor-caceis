```python
import os
import pandas as pd
import unicodedata

RAW_PATH = "../data/raw"
CLEAN_PATH = "../data/clean"


def clean_text(text):
    """Remove accents + clean filename"""
    text = unicodedata.normalize("NFKD", text).encode("ASCII", "ignore").decode("ASCII")
    text = text.replace(" ", "_").replace("-", "_")
    text = text.replace("__", "_")
    return text


# Walk through all folders
for root, dirs, files in os.walk(RAW_PATH):
    for file in files:
        if file.endswith((".xlsx", ".xls")):

            file_path = os.path.join(root, file)
            print(f"\nProcessing: {file}")

            try:
                # Load ALL sheets
                sheets = pd.read_excel(file_path, sheet_name=None)

                for sheet_name, df in sheets.items():

                    clean_file = clean_text(file.replace(".xlsx", "").replace(".xls", ""))
                    clean_sheet = clean_text(sheet_name)

                    # Limit length (Windows safety)
                    clean_file = clean_file[:40]
                    clean_sheet = clean_sheet[:30]

                    output_name = f"{clean_file}__{clean_sheet}.csv"
                    output_path = os.path.join(CLEAN_PATH, output_name)

                    df.to_csv(output_path, index=False, encoding="utf-8-sig")

                    print(f"Saved → {output_name}")

            except Exception as e:
                print(f"Error with {file}: {e}")
```

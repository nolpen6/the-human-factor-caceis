```python
import pandas as pd
import os
import sys

sys.path.append("../src")

from kpi_engineering import (
    build_absenteeism_kpi,
    build_learning_kpi,
    build_performance_kpi,
    build_engagement_kpi,
    build_talent_progression_kpi
)

CLEAN_PATH = "../data/clean"
OUTPUT_PATH = "../outputs"
```

```python
hr = pd.read_csv(os.path.join(
    CLEAN_PATH,
    "20250218__Stats_CACEIS_EAE_EP_18_02_2025__Database.csv"
))

performance = pd.read_csv(os.path.join(
    CLEAN_PATH,
    "20240222__CACEIS_Notes_evaluation_2023__Feuil1.csv"
))

absences = pd.read_csv(os.path.join(
    CLEAN_PATH,
    "20260121__Absenteisme__detail_affectatio__extract.csv"
))

training = pd.read_csv(os.path.join(
    CLEAN_PATH,
    "Training_Records_Unnamed__Final_CSV.csv"
))
```

```python
datasets = {
    "hr": hr,
    "performance": performance,
    "absences": absences,
    "training": training
}

for name, df in datasets.items():
    print(f"\n===== {name.upper()} =====")
    print("Shape:", df.shape)
    print(df.columns.tolist())
    display(df.head(3).T)
```

```python
def clean_cols(df):
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("'", "", regex=False)
        .str.replace("é", "e")
        .str.replace("è", "e")
        .str.replace("ê", "e")
        .str.replace("/", "_")
    )
    return df

hr = clean_cols(hr)
performance = clean_cols(performance)
absences = clean_cols(absences)
training = clean_cols(training)

print(hr.columns.tolist())
print(performance.columns.tolist())
print(absences.columns.tolist())
print(training.columns.tolist())
```

```python
HR_ID = "iug"
PERFORMANCE_ID = "iug"
PERFORMANCE_SCORE = "note"

ABSENCE_ID = "employee_code"
ABSENCE_DAYS = "jours_ouvres_absence"

TRAINING_ID = "employee_code"
TRAINING_HOURS = "total_training_hours"
TRAINING_SESSION = "session_id"
TRAINING_DATE = "seesion_start_date"
```

```python
performance[PERFORMANCE_SCORE] = pd.to_numeric(performance[PERFORMANCE_SCORE], errors="coerce")

performance_kpi = (
    performance
    .dropna(subset=[PERFORMANCE_ID])
    .groupby(PERFORMANCE_ID)
    .agg(
        avg_performance=(PERFORMANCE_SCORE, "mean"),
        performance_std=(PERFORMANCE_SCORE, "std"),
        performance_reviews=(PERFORMANCE_SCORE, "count")
    )
    .reset_index()
)
```

```python
# Fix missing std (1 review case)
performance_kpi["performance_std"] = performance_kpi["performance_std"].fillna(0)

# Flag low data reliability
performance_kpi["low_data_flag"] = performance_kpi["performance_reviews"] < 2
```

```python
performance_kpi["performance_consistency_score"] = (
    1 - performance_kpi["performance_std"].rank(pct=True)
)

performance_kpi["performance_consistency_score"] = (
    performance_kpi["performance_consistency_score"].fillna(1)
)

performance_kpi.head()
```

```python
absences[ABSENCE_DAYS] = pd.to_numeric(absences[ABSENCE_DAYS], errors="coerce")

absence_kpi = (
    absences
    .dropna(subset=[ABSENCE_ID])
    .groupby(ABSENCE_ID)
    .agg(
        absence_events=(ABSENCE_ID, "count"),
        absence_days=(ABSENCE_DAYS, "sum")
    )
    .reset_index()
)

absence_kpi["absence_events_score"] = absence_kpi["absence_events"].rank(pct=True)
absence_kpi["absence_days_score"] = absence_kpi["absence_days"].rank(pct=True)

absence_kpi["absenteeism_risk_score"] = (
    0.4 * absence_kpi["absence_events_score"] +
    0.6 * absence_kpi["absence_days_score"]
)

absence_kpi.head()
```

```python
training[TRAINING_HOURS] = pd.to_numeric(training[TRAINING_HOURS], errors="coerce")

training_kpi = (
    training
    .dropna(subset=[TRAINING_ID])
    .groupby(TRAINING_ID)
    .agg(
        training_count=(TRAINING_SESSION, "count"),
        training_hours=(TRAINING_HOURS, "sum")
    )
    .reset_index()
)

training_kpi["training_count_score"] = training_kpi["training_count"].rank(pct=True)
training_kpi["training_hours_score"] = training_kpi["training_hours"].rank(pct=True)

training_kpi["learning_intensity_score"] = (
    0.5 * training_kpi["training_count_score"] +
    0.5 * training_kpi["training_hours_score"]
)

training_kpi.head()
```

```python
talent_progression = (
    hr
    .dropna(subset=[HR_ID])
    .groupby(HR_ID)
    .agg(
        document_count=("nom_du_document", "count"),
        role=("libelle_emploi", "first"),
        contract_type=("type_contrat", "first"),
        country=("country", "first"),
        business_unit=("bu", "first")
    )
    .reset_index()
)

talent_progression["talent_progression_proxy"] = (
    talent_progression["document_count"].rank(pct=True)
)

talent_progression.head()
```

```python
hr_base = (
    hr
    .dropna(subset=[HR_ID])
    .drop_duplicates(subset=[HR_ID])
    [[HR_ID, "bu", "libelle_emploi", "type_contrat", "country", "libelle_organisation_niveau_07"]]
)

hr_kpi_table = hr_base.merge(
    performance_kpi,
    left_on=HR_ID,
    right_on=PERFORMANCE_ID,
    how="left"
)

hr_kpi_table = hr_kpi_table.merge(
    talent_progression[[HR_ID, "talent_progression_proxy"]],
    on=HR_ID,
    how="left"
)

hr_kpi_table.head()
```

```python
hr_kpi_table.to_csv("../outputs/hr_kpi_table.csv", index=False, encoding="utf-8-sig")
absence_kpi.to_csv("../outputs/absence_kpi_table.csv", index=False, encoding="utf-8-sig")
training_kpi.to_csv("../outputs/training_kpi_table.csv", index=False, encoding="utf-8-sig")

print("KPI tables saved successfully.")
```

Important limitation: HR/performance data uses `IUG`, while absenteeism and training use `Employee Code`. Since these identifiers do not appear to match directly, we avoid unsafe merges. Absenteeism and learning KPIs are therefore calculated in separate employee-level tables, while the main HR KPI table combines HR and performance data through `IUG`.
